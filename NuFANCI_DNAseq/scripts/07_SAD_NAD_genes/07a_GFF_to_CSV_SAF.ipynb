```python
import gffpandas.gffpandas as gffpd
```

```python
annotation = gffpd.read_gff3('chm13.draft_v2.0.gene_annotation.gff3')
```

```python
filtered_df = annotation.filter_feature_of_type(['gene'])
attr_to_columns = filtered_df.attributes_to_columns()
```

```python
#attr_to_columns.to_tsv('chm13.draft_v2.0.gene_annotation.onlygenes.tsv')
attr_to_columns.to_csv('chm13.draft_v2.0.gene_annotation.onlygenes.csv')
```
