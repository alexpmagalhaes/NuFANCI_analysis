# Select proteins for GO enrichment

Reads the NPM1 vs SRRM2 sheet of Supplementary Table S3 and writes the filtered protein list used for GO enrichment (`filtered_results_GreyDots_plus5values.csv`).

**Inputs** (in the working directory): `Table_S3_NSPKL_FANCI_Proteomics_250813a.xlsx` (Supplementary Table S3).

Original notebook: `filtered_results_GreyDots_plus5values.ipynb`.

```python
from __future__ import print_function
import sys
import pandas as pd
import csv
```

```python
# --- Configuration ---
file_path = 'Table_S3_NSPKL_FANCI_Proteomics_250813a.xlsx'
sheet_name = 'NPM1 vs SRRM2 (Ab sort)'
significance_col = 'NPKL_SRRM2_Ab_FANCI_vs_NUCL_NPM1_AB_FANCI_significant'
```

```python
# --- Step 1 & 2: Import the specific sheet and skip the first row ---
# The 'skiprows=1' argument tells pandas to skip the first row of the sheet.
print(f"Reading data from '{file_path}' | Sheet: '{sheet_name}'...")
df = pd.read_excel(file_path, sheet_name=sheet_name, skiprows=1)
print(f"Successfully loaded data. Original shape: {df.shape}")
print("-" * 30)

# --- Step 3: Filter rows by the 'significant' column ---
# We keep rows where the value in the significance column is False.
# Note: Pandas often reads 'FALSE' from Excel as a boolean False.
# If this step results in an empty table, try changing '== False' to '== "FALSE"'
print(f"Filtering rows where '{significance_col}' is FALSE...")
df_filtered_significance = df[df[significance_col] == False].copy()
print(f"Shape after significance filter: {df_filtered_significance.shape}")
print("-" * 30)

# --- Step 4: Filter rows based on the number of values in the last 6 columns ---
# We identify the last 6 columns, count the non-missing values for each row,
# and keep rows where the count is 5 or more.
print("Filtering rows with 5 or more expression values in the last 6 columns...")

# Select the last 6 columns
expression_columns = df_filtered_significance.iloc[:, -6:]

# Create a boolean mask for rows that meet the condition
# .count(axis=1) counts non-null values across each row
row_mask = expression_columns.count(axis=1) >= 5

# Apply the mask to the dataframe
final_df = df_filtered_significance[row_mask]

print(f"Shape after expression value filter: {final_df.shape}")
print("-" * 30)


# --- Display Results ---
print("\nFinal Filtered Data (First 10 rows):")
print(final_df.head(10))

# --- Optional: Save the result to a new file ---
# Uncomment the line below to save the filtered data to a CSV or Excel file.
final_df.to_csv('filtered_results.csv', index=False)
final_df.to_excel('filtered_results.xlsx', index=False)
print("\nFiltered data saved to 'filtered_results.csv'")
```
